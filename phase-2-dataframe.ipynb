##  Data frame

## when you combine a series one two more is a dataframe

## creating a Dataframe

```python
import pandas as pd
import numpy as np
```

## creating dataframe

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df=pd.DataFrame(data)
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2=pd.DataFrame(data_list)
column=['Name','Age','City','Salary']
df2=pd.DataFrame(data_list,columns=column)
df2
```

### Selection and indexing

```python
df2['Name']
```

```python
df2[['Name','City']]
```

## creatinng a new column

```python
df2['Designation']=["Doctor","Eng","Doctor","Plumber"]
df2
```

## remove a column

```python
df2.drop("Designation",axis=1,inplace=True)
df2
```

```python
df2.drop(["City","Salary"],axis=1)
```

```python
df2.drop(0,axis=0)
```
